```python
def binary_search(arr, left, right, target):
    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1
```

### W7A2

```python
def count_occurrences(arr, x):
    count = 0

    for item in arr:
        if item == x:
            count += 1

    return count
```

### W7A3

```python
def bubble_sort(arr):
    n = len(arr)

    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

    return arr
```

### W7A4

```python
def most_frequent(arr):
    max_count = 0
    result = arr[0]

    for item in arr:
        count = arr.count(item)

        if count > max_count:
            max_count = count
            result = item

    return result
```

### W7A5

```python
def count_pairs(arr, x):
    count = 0
    n = len(arr)

    for i in range(n):
        for j in range(i + 1, n):
            if arr[i] + arr[j] == x:
                count += 1

    return count
```

### W7A6

```python
def flatten_list(nested_list):
    result = []

    for sublist in nested_list:
        for item in sublist:
            result.append(item)

    return result
```

### W7A7

```python
def longest_increasing_subsequence(nums):
    n = len(nums)

    if n == 0:
        return 0

    dp = [1] * n

    for i in range(n):
        for j in range(i):
            if nums[i] > nums[j]:
                dp[i] = max(dp[i], dp[j] + 1)

    return max(dp)
```
