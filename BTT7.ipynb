```python
def binary_search(arr, left, right, target):
    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1
```

### W7A3

```python
def bubble_sort(arr):
    n = len(arr)

    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

    return arr
```

### W7A7

```python
def longest_increasing_subsequence(nums):
    n = len(nums)

    if n == 0:
        return 0

    dp = [1] * n

    for i in range(n):
        for j in range(i):
            if nums[i] > nums[j]:
                dp[i] = max(dp[i], dp[j] + 1)

    return max(dp)
```

### W7A8

```python
def merge_sort(arr):

    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)


def merge(a, b):

    res = []

    i = 0
    j = 0

    while i < len(a) and j < len(b):

        if a[i][0] <= b[j][0]:
            res.append(a[i])
            i += 1
        else:
            res.append(b[j])
            j += 1

    while i < len(a):
        res.append(a[i])
        i += 1

    while j < len(b):
        res.append(b[j])
        j += 1

    return res


class MinHeap:

    def __init__(self):
        self.heap = []

    def push(self, val):

        self.heap.append(val)

        self.up(len(self.heap) - 1)

    def up(self, idx):

        while idx > 0:

            parent = (idx - 1) // 2

            if self.heap[parent][0] <= self.heap[idx][0]:
                break

            self.heap[parent], self.heap[idx] = \
                self.heap[idx], self.heap[parent]

            idx = parent

    def pop(self):

        if len(self.heap) == 0:
            return

        last = len(self.heap) - 1

        self.heap[0], self.heap[last] = \
            self.heap[last], self.heap[0]

        self.heap.pop()

        self.down(0)

    def down(self, idx):

        n = len(self.heap)

        while True:

            smallest = idx

            left = idx * 2 + 1
            right = idx * 2 + 2

            if left < n and \
               self.heap[left][0] < self.heap[smallest][0]:

                smallest = left

            if right < n and \
               self.heap[right][0] < self.heap[smallest][0]:

                smallest = right

            if smallest == idx:
                break

            self.heap[idx], self.heap[smallest] = \
                self.heap[smallest], self.heap[idx]

            idx = smallest

    def top(self):

        if len(self.heap) == 0:
            return None

        return self.heap[0]

    def empty(self):
        return len(self.heap) == 0


def minInterval(intervals, queries):
    intervals = merge_sort(intervals)

    q = []

    for i in range(len(queries)):
        q.append([queries[i], i])

    q = merge_sort(q)

    res = [-1] * len(queries)

    heap = MinHeap()

    i = 0
    n = len(intervals)

    for item in q:

        query = item[0]
        idx = item[1]

        while i < n and intervals[i][0] <= query:

            l = intervals[i][0]
            r = intervals[i][1]

            size = r - l + 1

            heap.push([size, r])

            i += 1

        while not heap.empty() and heap.top()[1] < query:
            heap.pop()

        if not heap.empty():
            res[idx] = heap.top()[0]

    return res


intervals = [[1,4],[2,4],[3,6],[4,4]]
queries = [2,3,4,5]

print(minInterval(intervals, queries))
```

### W7A9

```python
class Node:

    def __init__(self, val):

        self.val = val

        self.left = None
        self.right = None

        self.left_count = 0
        self.count = 1


def insert(node, val):

    smaller = 0

    while True:

        if val == node.val:

            node.count += 1

            smaller += node.left_count

            return smaller
        elif val < node.val:

            node.left_count += 1

            if node.left is None:
                node.left = Node(val)
                return smaller

            node = node.left
        else:

            smaller += node.left_count + node.count

            if node.right is None:
                node.right = Node(val)
                return smaller

            node = node.right


def countSmaller(nums):

    n = len(nums)

    if n == 0:
        return []

    res = [0] * n

    root = Node(nums[n - 1])
    for i in range(n - 2, -1, -1):

        res[i] = insert(root, nums[i])

    return res


nums = [5,2,6,1]

print(countSmaller(nums))
```

### W7A10

```python
def binary_search(arr, target):
    left = 0
    right = len(arr)

    while left < right:
        mid = (left + right) // 2

        if arr[mid] < target:
            left = mid + 1
        else:
            right = mid

    return left


def insert_sorted(arr, value):
    pos = binary_search(arr, value)

    arr.append(0)

    i = len(arr) - 1
    while i > pos:
        arr[i] = arr[i - 1]
        i -= 1

    arr[pos] = value


def count_smaller(nums):
    sorted_list = []
    result = []

    i = len(nums) - 1

    while i >= 0:
        pos = binary_search(sorted_list, nums[i])
        result.append(pos)

        insert_sorted(sorted_list, nums[i])

        i -= 1

    result.reverse()
    return result
```

### W7A16

```python
def min_delete(arr):
    count = {}

    for x in arr:
        if x in count:
            count[x] += 1
        else:
            count[x] = 1

    delete_count = 0

    for x in count:
        freq = count[x]

        if freq < x:
            delete_count += freq
        elif freq > x:
            delete_count += freq - x

    return delete_count

arr = list(map(int, input().split()))
print(min_delete(arr))
```

```python
def majority_element(arr):
    candidate = None
    count = 0
    for num in arr:
        if count == 0:
            candidate = num
            count = 1
        elif num == candidate:
            count += 1
        else:
            count -= 1
    freq = 0

    for num in arr:
        if num == candidate:
            freq += 1

    if freq > len(arr) // 2:
        return candidate
    else:
        return "No"

arr = list(map(int, input().split()))

print(majority_element(arr))
```

```python
s = input()

stack = []

valid = True

for c in s:
    if c == '(' or c == '[' or c == '{':
        stack.append(c)

    else:
        if len(stack) == 0:
            valid = False
            break

        top = stack[-1]
        stack.pop()

        if c == ')' and top != '(':
            valid = False

        elif c == ']' and top != '[':
            valid = False

        elif c == '}' and top != '{':
            valid = False

if len(stack) != 0:
    valid = False

if valid:
    print("Valid")
else:
    print("Invalid")
```

### W7A28

```python
def is_prime(n):
    if n < 2:
        return False

    i = 2
    while i * i <= n:
        if n % i == 0:
            return False
        i += 1

    return True


N = int(input())

best_start = 0
best_len = 0

current_start = 0
current_len = 0

for i in range(1, N + 1):
    if not is_prime(i):
        if current_len == 0:
            current_start = i

        current_len += 1

        if current_len > best_len:
            best_len = current_len
            best_start = current_start
    else:
        current_len = 0

for i in range(best_start, best_start + best_len):
    print(i, end=" ")
```
