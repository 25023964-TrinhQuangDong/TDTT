```python
class Date:
    def __init__(self, dd, mm, yy):
        self.day = dd
        self.month = mm
        self.year = yy
    
    def checkValids(self):
        maxday = -1
        match self.month:
            case 1|3|5|7|8|10|12:
                maxday = 31
            case 4|6|9|11:
                maxday = 30
            case 2:
                if (self.year%4 == 0 and self.year%100 != 0) or self.year %400 == 0:
                    maxday = 29
                else:
                    maxday = 28
        return 1 <= self.day <= maxday and self.year > 0 
    def getNextday(self):
        if(self.checkValids()):
            day_1 = Date(self.day +1, self.month , self.year)
            if day_1.checkValids():
                return day_1
            day_1 = Date(1, self.month +1, self.year)
            if day_1.checkValids():
                return day_1
            return Date(1, 1, self.year +1)
    def __str__(self):
        return f"{self.day:02d}/{self.month:02d}/{self.year}"
    

d,m,y = map(int, input().split("/"))
day_1 = Date(d, m, y)
print(day_1.checkValids())

day_2 = day_1.getNextday()
print(day_2)
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def is_origin(self):
        return self.x == 0 and self.y == 0

    def on_x_axis(self):
        return self.y == 0

    def on_y_axis(self):
        return self.x == 0

    def distance_to_origin(self):
        return (self.x * self.x + self.y * self.y) ** 0.5


x, y = map(float, input().split())

p = Point(x, y)

if p.is_origin():
    print("M la goc toa do")
else:
    print("M khong phai goc toa do")

if p.on_x_axis():
    print("M nam tren truc hoanh")
else:
    print("M khong nam tren truc hoanh")

if p.on_y_axis():
    print("M nam tren truc tung")
else:
    print("M khong nam tren truc tung")

print(f"{p.distance_to_origin():.2f}")
```

```python
class ShoppingCart:
    def __init__(self):
        self.products = []

    def add_product(self, name, price):
        product = {
            "name": name,
            "price": price
        }
        self.products.append(product)

    def remove_product(self, name):
        found = False

        for i in range(len(self.products)):
            if self.products[i]["name"] == name:
                del self.products[i]
                found = True
                break

        if found:
            print("Da xoa san pham")
        else:
            print("Khong tim thay san pham")

    def is_empty(self):
        return len(self.products) == 0

    def total_price(self):
        total = 0

        for p in self.products:
            total += p["price"]

        return total

    def show_products(self):
        if self.is_empty():
            print("Gio hang rong")
        else:
            for p in self.products:
                print(p["name"], "-", f"{p['price']:.2f}")

    def clear_cart(self):
        self.products = []
        print("Da xoa toan bo gio hang")


cart = ShoppingCart()

while True:
    print("\n===== MENU =====")
    print("2. Them san pham")
    print("3. Xoa san pham")
    print("4. Kiem tra gio hang rong")
    print("5. Tong tien gio hang")
    print("6. Hien thi san pham")
    print("7. Xoa toan bo gio hang")

    choice = int(input("Nhap lua chon: "))

    if choice == 2:
        name = input("Ten san pham: ")
        price = float(input("Gia: "))
        cart.add_product(name, price)

    elif choice == 3:
        name = input("Nhap ten san pham can xoa: ")
        cart.remove_product(name)

    elif choice == 4:
        if cart.is_empty():
            print("Gio hang rong")
        else:
            print("Gio hang co san pham")

    elif choice == 5:
        print(f"Tong tien: {cart.total_price():.2f}")

    elif choice == 6:
        cart.show_products()

    elif choice == 7:
        cart.clear_cart()

    ans = input("Co muon thoat chuong trinh khong? ")

    if ans == "yes":
        break
```

```python
class Fraction:
    def __init__(self, num, den):
        self.num = num
        self.den = den
        self.reduce()

    def gcd(self, a, b):
        while b != 0:
            temp = a % b
            a = b
            b = temp
        return a

    def reduce(self):
        g = self.gcd(abs(self.num), abs(self.den))

        self.num //= g
        self.den //= g

        if self.den < 0:
            self.num *= -1
            self.den *= -1

    def add(self, other):
        num = self.num * other.den + other.num * self.den
        den = self.den * other.den
        return Fraction(num, den)

    def sub(self, other):
        num = self.num * other.den - other.num * self.den
        den = self.den * other.den
        return Fraction(num, den)

    def mul(self, other):
        num = self.num * other.num
        den = self.den * other.den
        return Fraction(num, den)

    def div(self, other):
        num = self.num * other.den
        den = self.den * other.num
        return Fraction(num, den)

    def __str__(self):
        return str(self.num) + "/" + str(self.den)

a, b, op, c, d = input().split()
a = int(a)
b = int(b)
c = int(c)
d = int(d)
f1 = Fraction(a, b)
f2 = Fraction(c, d)
if op == '+':
    result = f1.add(f2)
elif op == '-':
    result = f1.sub(f2)
elif op == '*':
    result = f1.mul(f2)
elif op == '/':
    result = f1.div(f2)
print(result)
```

```python
def my_sqrt(x):
    t = x
    for i in range(100):
        t = (t + x / t) / 2
    return t
def my_abs(x):
    if x < 0:
        return -x
    return x
def gt(n):
    kq = 1
    for i in range(1, n + 1):
        kq *= i
    return kq
def cos_taylor(x):
    s = 1
    dau = -1

    for i in range(2, 20, 2):
        s += dau * (x ** i) / gt(i)
        dau *= -1
    return s
def arccos_custom(c):
    pi = 3.141592653589793
    left = 0
    right = pi
    for i in range(100):
        mid = (left + right) / 2

        if cos_taylor(mid) > c:
            left = mid
        else:
            right = mid
    return (left + right) / 2
def rad_to_deg(x):
    pi = 3.141592653589793
    return x * 180 / pi
A = list(map(float, input().split()))
B = list(map(float, input().split()))
C = list(map(float, input().split()))
D = list(map(float, input().split()))
ABx = B[0] - A[0]
ABy = B[1] - A[1]
ABz = B[2] - A[2]
ACx = C[0] - A[0]
ACy = C[1] - A[1]
ACz = C[2] - A[2]
BCx = C[0] - B[0]
BCy = C[1] - B[1]
BCz = C[2] - B[2]
BDx = D[0] - B[0]
BDy = D[1] - B[1]
BDz = D[2] - B[2]
n1x = ABy * ACz - ABz * ACy
n1y = ABz * ACx - ABx * ACz
n1z = ABx * ACy - ABy * ACx
n2x = BCy * BDz - BCz * BDy
n2y = BCz * BDx - BCx * BDz
n2z = BCx * BDy - BCy * BDx
dot = n1x * n2x + n1y * n2y + n1z * n2z
len1 = my_sqrt(n1x * n1x + n1y * n1y + n1z * n1z)
len2 = my_sqrt(n2x * n2x + n2y * n2y + n2z * n2z)
cosphi = dot / (len1 * len2)
if cosphi > 1:
    cosphi = 1
if cosphi < -1:
    cosphi = -1
if cosphi < 0:
    cosphi = -cosphi
rad = arccos_custom(cosphi)
deg = rad_to_deg(rad)
print(round(deg, 2))
```

```python
class Triangle:
    def __init__(self, intEdge1, intEdge2, intEdge3):
        self.edge1 = intEdge1
        self.edge2 = intEdge2
        self.edge3 = intEdge3

    def my_sqrt(self, x):
        if x == 0:
            return 0
        t = x
        for i in range(100):
            t = (t + x / t) / 2
        return t
    def getArea(self):
        a = self.edge1
        b = self.edge2
        c = self.edge3
        if a + b <= c or a + c <= b or b + c <= a:
            return -1
        p = (a + b + c) / 2
        s = p * (p - a) * (p - b) * (p - c)
        return self.my_sqrt(s)
    
a, b, c = map(int, input().split())
tg = Triangle(a, b, c)
area = tg.getArea()
if area == -1:
    print(-1)
else:
    print(format(area, ".2f"))
```
