```python
def merge_sorted(a, b):
    i = j = 0
    res = []

    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            res.append(a[i])
            i += 1
        else:
            res.append(b[j])
            j += 1

    while i < len(a):
        res.append(a[i])
        i += 1

    while j < len(b):
        res.append(b[j])
        j += 1

    return res

a = list(map(int, input().split()))
b = list(map(int, input().split()))

print(*merge_sorted(a, b))
```

### W9A2

```python
def smallest_greater_than_k(matrix, n, m, k):
    ans = None

    for i in range(n):
        for j in range(m):
            if matrix[i][j] > k:
                if ans is None or matrix[i][j] < ans:
                    ans = matrix[i][j]

    return ans


n, m = map(int, input().split())

matrix = []
for _ in range(n):
    matrix.append(list(map(int, input().split())))

k = int(input())

print(smallest_greater_than_k(matrix, n, m, k))
```

### W9A3

```python
def smallest_by_one_swap(num):
    s = list(num)
    n = len(s)

    for i in range(n):
        pos = i

        for j in range(n - 1, i, -1):
            if s[j] < s[pos]:
                if i == 0 and s[j] == '0':
                    continue
                pos = j

        if pos != i:
            s[i], s[pos] = s[pos], s[i]
            break

    return ''.join(s)


num = input().strip()
print(smallest_by_one_swap(num))
```

### W9A4

```python
def abbreviation(a, b):
    n = len(a)
    m = len(b)

    dp = [[False] * (m + 1) for _ in range(n + 1)]
    dp[0][0] = True

    for i in range(n):
        for j in range(m + 1):

            if not dp[i][j]:
                continue

            if a[i].islower():
                dp[i + 1][j] = True

            if j < m and a[i].upper() == b[j]:
                dp[i + 1][j + 1] = True

    return "YES" if dp[n][m] else "NO"


q = int(input())

for _ in range(q):
    a = input().strip()
    b = input().strip()

    print(abbreviation(a, b))
```

```python
def is_subarray(list1, list2):
    n = len(list1)
    m = len(list2)

    for i in range(m - n + 1):
        if list2[i:i + n] == list1:
            return True

    return False


list1 = list(map(int, input().split()))
list2 = list(map(int, input().split()))

print("YES" if is_subarray(list1, list2) else "NO")
```

### W9A6

```python
def next_permutation(a):
    n = len(a)

    i = n - 2
    while i >= 0 and a[i] > a[i + 1]:
        i -= 1

    if i == -1:
        return sorted(a)

    j = n - 1
    while a[j] < a[i]:
        j -= 1

    a[i], a[j] = a[j], a[i]

    a[i + 1:] = reversed(a[i + 1:])

    return a


n = int(input())
a = list(map(int, input().split()))

print(*next_permutation(a))
```

### W9A7

```python
def canPlaceFlowers(flowerbed, k):
    n = len(flowerbed)

    for i in range(n):

        if flowerbed[i] == 0:

            left_empty = (i == 0 or flowerbed[i - 1] == 0)
            right_empty = (i == n - 1 or flowerbed[i + 1] == 0)

            if left_empty and right_empty:
                flowerbed[i] = 1
                k -= 1

                if k == 0:
                    return True

    return k <= 0


flowerbed = list(map(int, input().split()))
k = int(input())

print(canPlaceFlowers(flowerbed, k))
```

### W9A8

```python
def is_path_crossing(moves):
    x, y = 0, 0

    visited = set()
    visited.add((0, 0))

    for move in moves:

        if move == 'R':
            x += 1

        elif move == 'L':
            x -= 1

        elif move == 'U':
            y += 1

        elif move == 'D':
            y -= 1

        if (x, y) in visited:
            return True

        visited.add((x, y))

    return False


moves = input().strip()

print(is_path_crossing(moves))
```

### W9A9

```python
def Median(nums):
    a, b, c, d, e = nums

    if a > b:
        a, b = b, a

    if c > d:
        c, d = d, c
    if b > d:
        a, c = c, a
        b, d = d, b

    if b < c:

        if c < e:
            return c

        elif b < e:
            return e

        else:
            return b

    else:

        if b < e:
            return b

        elif c < e:
            return e

        else:
            return c
        
nums = list(map(int, input().split()))
print(Median(nums))
```

### W9A10

```python
def findRadius(houses, heaters):

    for i in range(len(houses)):
        min_idx = i

        for j in range(i + 1, len(houses)):
            if houses[j] < houses[min_idx]:
                min_idx = j

        houses[i], houses[min_idx] = houses[min_idx], houses[i]

    for i in range(len(heaters)):
        min_idx = i

        for j in range(i + 1, len(heaters)):
            if heaters[j] < heaters[min_idx]:
                min_idx = j

        heaters[i], heaters[min_idx] = heaters[min_idx], heaters[i]

    ans = 0

    for house in houses:

        nearest = 10**18

        for heater in heaters:

            dist = house - heater
            if dist < 0:
                dist = -dist

            if dist < nearest:
                nearest = dist

        if nearest > ans:
            ans = nearest

    return ans
```

### W9A11

```python
def factorial(n):

    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)
```

```python
def sum_to_n(n):

    if n == 1:
        return 1

    return n + sum_to_n(n - 1)
```

### W9A13

```python
def power(base, exp):

    if exp == 0:
        return 1

    half = power(base, exp // 2)

    if exp % 2 == 0:
        return half * half

    return base * half * half
```

### W9A14

```python
def find_more_than_x(arr, x):

    result = []

    used = []

    for i in range(len(arr)):

        existed = False

        for v in used:
            if v == arr[i]:
                existed = True
                break

        if existed:
            continue

        count = 0

        for j in range(len(arr)):
            if arr[j] == arr[i]:
                count += 1

        if count > x:
            result.append(arr[i])

        used.append(arr[i])

    if len(result) == 0:
        return "INVALID"

    return result
```

### W9A15

```python
def check_sequence(arr):

    increasing = True
    decreasing = True

    for i in range(1, len(arr)):

        if arr[i] <= arr[i - 1]:
            increasing = False

        if arr[i] >= arr[i - 1]:
            decreasing = False

    if increasing or decreasing:
        return True

    return False
```
