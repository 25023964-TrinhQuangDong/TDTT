### W10A3

```python
try:
    n = int(input())
    visited = []
    while n != 1:
        if n in visited:
            print("False")
            break
        visited.append(n)
        s = 0
        temp = n
        while temp > 0:
            d = temp % 10
            s = s + d * d
            temp = temp // 10
        n = s
    else:
        print("True")
except ValueError:
    print("Loi nhap lieu")
```

### W10A8

```python
filename = input()
try:
    n = len(filename)
    if n >= 4:
        ext = ""
        i = n - 4
        while i < n:
            ext = ext + filename[i]
            i = i + 1
        if ext == ".txt" or ext == ".zip":
            print("Doc file thanh cong")
        else:
            raise ValueError
    else:
        raise ValueError
except:
    print("File khong hop le")
```

### W10A10

```python
try:
    N = int(input())
    s = input() + " "
    a = []
    so = ""
    i = 0
    while i < len(s):
        if s[i] != " ":
            so = so + s[i]
        else:
            a.append(int(so))
            so = ""
        i = i + 1
    hop_le = True
    i = 0
    while i < N:
        j = i + 1
        while j < N:
            if a[i] == a[j]:
                hop_le = False
            j = j + 1
        i = i + 1
    if hop_le:
        tong = 0
        i = 0
        while i < N:
            tong = tong + a[i]
            i = i + 1
        print(tong)
    else:
        print("Mang khong hop le")
except:
    print("Mang khong hop le")
```
